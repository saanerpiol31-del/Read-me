```python
#Create a program that asks a user to enter a password. The password must be 8 to 15 character long
password = input("Create a password: ")

length = len(password)

if length < 8:
    print("Password is too short.")
elif length > 15:
    print("Password is too long.")
else:
    print("Password is valid!")
```

```python
#Create a program that checks if a person is eligible for a loan
credit = int(input("Enter credit score: "))
income = float(input("Enter income: "))
job = float(input("Enter years at job: "))

if credit >= 700 and income >= 30000 and job >= 2:
    print("Loan Approved")
else:
    print("Loan Denied")
```
