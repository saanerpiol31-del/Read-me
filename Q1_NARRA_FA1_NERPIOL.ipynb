```python
#Finding the distance betweeen two points
import math

x1 =float(input("enter x1: "))
x2 =float(input("enter x2: "))
y1 =float(input("enter y1: "))
y2 =float(input("enter y2: "))

distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1 ,2))

print(f"The distance is:  {distance}")


#When using a library it is more practical because it is easier and automatic. You can also see where you got it wrong to help you.
```
